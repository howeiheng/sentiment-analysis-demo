## Step 2: Load the Sentiment Analysis Model

Now we'll load a powerful, pre-trained AI model called **DistilBERT**. This model is:
- Fast and efficient
- Highly accurate
- Trained on millions of text examples
- Works great in Google Colab

This might take a moment the first time it runs.

```python
from transformers import pipeline
import pandas as pd

# Load the sentiment analysis model
print("Loading model (this may take a moment on first run)...")
sentiment_analyzer = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

print("✓ Model loaded successfully!")
print("Ready to analyze sentiment!")
```

## Step 3: Analyze Your Text!

Type in any text you want to analyze when the cell runs!

### Try these examples:
- "I love this! This is amazing!"
- "This is terrible and I hate it."
- "The weather is nice today."
- "Best day ever!"
- "I'm so disappointed and frustrated."

```python
# Get text input from the user
text_to_analyze = input("📝 Enter the text you want to analyze: ")

if text_to_analyze.strip() == "":
    print("⚠️  Please enter some text to analyze!")
else:
    # Analyze the sentiment
    result = sentiment_analyzer(text_to_analyze)
    
    # Extract and display results
    sentiment_label = result[0]['label']
    confidence_score = round(result[0]['score'] * 100, 2)
    
    print(f"\n📝 Text: \"{text_to_analyze}\"")
    print(f"\n😊 Sentiment: {sentiment_label}")
    print(f"📊 Confidence Score: {confidence_score}%")
    
    # Show emoji based on sentiment
    if sentiment_label == 'POSITIVE':
        print("\n🎉 This text has a positive vibe!")
    elif sentiment_label == 'NEGATIVE':
        print("\n😢 This text has a negative vibe.")
    else:
        print("\n😐 This text is neutral.")
```

## Try More Examples!

### How does this work?

This notebook uses a **machine learning model** called **DistilBERT** that has been trained on thousands of examples of text. It learned patterns that help it recognize whether text is positive, negative, or neutral.

The **confidence score** tells you how sure the model is about its answer:
- 90% = Very confident
- 60% = Somewhat confident
- 51% = Just slightly confident

## Challenge: Analyze Multiple Texts

Run the cell below to analyze several texts at once!

```python
# List of texts to analyze
texts = [
    "This product is absolutely fantastic!",
    "Worst experience ever, very disappointed.",
    "The weather is cloudy today.",
    "I'm so happy and excited!",
    "This is boring and waste of time."
]

print("Analyzing multiple texts...\n")
print("=" * 70)

for text in texts:
    result = sentiment_analyzer(text)
    label = result[0]['label']
    score = round(result[0]['score'] * 100, 2)
    
    emoji = "🎉" if label == "POSITIVE" else "😢" if label == "NEGATIVE" else "😐"
    print(f"{emoji} {label:8} ({score}%): {text}")

print("=" * 70)
```

## How to Prepare Your File

Your Excel file (`.xlsx`) should have:

1. **A column called `index`** - This is a unique identifier for each row (can be numbers, IDs, etc.)
2. **A column called `text`** - This contains the text you want to analyze for sentiment
3. **Any other columns** - These will be kept in the output file

### Example of a valid file:

| index | text | author | date |
|-------|------|--------|------|
| 1 | I love this product! | Alice | 2024-01-01 |
| 2 | This is terrible | Bob | 2024-01-02 |
| 3 | It's okay, nothing special | Carol | 2024-01-03 |

### Output file will have:

| index | text | author | date | sentiment | sentiment_score |
|-------|------|--------|------|-----------|---------------|
| 1 | I love this product! | Alice | 2024-01-01 | POSITIVE | 0.9998 |
| 2 | This is terrible | Bob | 2024-01-02 | NEGATIVE | 0.9994 |
| 3 | It's okay, nothing special | Carol | 2024-01-03 | POSITIVE | 0.5234 |

## Step 1: Choose Your Data Source

Run the cell below to choose whether you want to use your own file or a demo file.

```python
from google.colab import files
import os

print("\n📋 How would you like to proceed?")
print("\nOption 1: Upload your own Excel file")
print("Option 2: Use a demo dataset")
print("\nEnter 1 or 2:")

choice = input().strip()

if choice == "1":
    print("\n📁 Choose your Excel file to upload...")
    uploaded_files = files.upload()
    
    if uploaded_files:
        uploaded_filename = list(uploaded_files.keys())[0]
        print(f"\n✓ File uploaded: {uploaded_filename}")
        use_demo = False
    else:
        print("No file uploaded! Using demo data instead.")
        use_demo = True
        
elif choice == "2":
    print("\n📊 Using demo dataset...")
    use_demo = True
else:
    print("\n❌ Invalid choice. Using demo data instead.")
    use_demo = True

# If using demo data, create a sample dataframe
if use_demo:
    print("\n✓ Demo dataset created!")
    demo_data = {
        'index': [1, 2, 3, 4, 5, 6, 7, 8],
        'text': [
            'I absolutely love this product! Best purchase ever!',
            'This is terrible and a complete waste of money.',
            'The weather today is nice and sunny.',
            'I am extremely disappointed with this service.',
            'Amazing experience! Highly recommend to everyone!',
            'Worst customer service I have ever experienced.',
            'This item is okay, nothing special really.',
            'Fantastic! Could not be happier with my purchase!'
        ],
        'source': [
            'review',
            'review',
            'social_media',
            'review',
            'review',
            'review',
            'social_media',
            'review'
        ]
    }
    df = pd.DataFrame(demo_data)
    uploaded_filename = 'demo_data'
else:
    df = pd.read_excel(uploaded_filename)

print(f"\n📊 File Info:")
print(f"  - Total rows: {len(df)}")
print(f"  - Columns: {list(df.columns)}")

# Check for required columns
if 'index' not in df.columns:
    print("\n⚠️  WARNING: Column 'index' not found!")
if 'text' not in df.columns:
    print("\n⚠️  WARNING: Column 'text' not found!")

print(f"\n📄 First few rows:")
print(df.head())
```

## Step 2: Analyze Sentiment

This will analyze the sentiment for each row in your data. This might take a few moments depending on how many rows you have.

⏱️ **Note:** For large files (1000+ rows), this may take several minutes. DistilBERT is optimized for speed, so it should be faster than heavier models.

```python
print("🔄 Analyzing sentiment... This may take a moment...\n")

# Initialize lists to store results
sentiments = []
scores = []

# Analyze each text
total_rows = len(df)
for idx, row in df.iterrows():
    text = str(row['text'])  # Convert to string in case of NaN values
    
    # Analyze sentiment
    result = sentiment_analyzer(text)
    sentiment_label = result[0]['label']
    sentiment_score = round(result[0]['score'], 4)
    
    sentiments.append(sentiment_label)
    scores.append(sentiment_score)
    
    # Print progress every 5 rows
    if (idx + 1) % 5 == 0 or (idx + 1) == total_rows:
        print(f"  Progress: {idx + 1}/{total_rows} rows analyzed...")

# Add sentiment columns to dataframe
df['sentiment'] = sentiments
df['sentiment_score'] = scores

print(f"\n✓ Sentiment analysis complete!")
print(f"\n📊 Results summary:")
print(df['sentiment'].value_counts())
```

## Step 3: Preview Results

Here's a preview of your data with the new sentiment columns added.

```python
print("📄 First 10 rows with sentiment analysis:")
print(df.head(10))

print("\n\n📈 Sentiment Distribution:")
print(df['sentiment'].value_counts())

print("\n\n📊 Average Sentiment Score by Sentiment:")
print(df.groupby('sentiment')['sentiment_score'].mean().round(4))
```

## Step 4: Download Your Results

Run the cell below to download your file with the sentiment analysis results!

```python
# Create output filename by adding "_with_sentiment" suffix
# Example: "data.xlsx" becomes "data_with_sentiment.xlsx"
if uploaded_filename == 'demo_data':
    output_filename = "demo_data_with_sentiment.xlsx"
else:
    original_name = uploaded_filename.rsplit('.', 1)[0]  # Remove .xlsx
    output_filename = f"{original_name}_with_sentiment.xlsx"

# Save to Excel
df.to_excel(output_filename, index=False)

print(f"✓ File saved as: {output_filename}")
print(f"\n⬇️ Click the download button below or in the Files panel to download your file.")

# Download the file
files.download(output_filename)

print(f"\n🎉 Done! Your sentiment analysis is ready to download.")
```
